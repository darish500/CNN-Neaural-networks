```python
img_NZ_bgr = cv2.imread("/content/lion.jpg", cv2.IMREAD_COLOR)
b, g, r = cv2.split(img_NZ_bgr)

# Show the channels
plt.figure(figsize=[20, 5])

plt.subplot(141);plt.imshow(r, cmap="gray");plt.title("Red Channel")
plt.subplot(142);plt.imshow(g, cmap="gray");plt.title("Green Channel")
plt.subplot(143);plt.imshow(b, cmap="gray");plt.title("Blue Channel")

# Merge the individual channels into a BGR image
imgMerged = cv2.merge((b, g, r))
# Show the merged output
plt.subplot(144)
plt.imshow(imgMerged[:, :, ::-1])
plt.title("Merged Output")
```

```python
import cv2 as cv
import matplotlib.pyplot as plt

# Load the image
img = cv.imread("/content/lion.jpg")
# Resize the image (set width to 400px and height proportionally)
width = 18

aspect_ratio = width / img.shape[1]
height = int(img.shape[0] * aspect_ratio)
resized_img = cv.resize(img, (width, height))

# Convert the resized image from BGR to RGB (for matplotlib)
img_rgb = cv.cvtColor(resized_img, cv.COLOR_BGR2RGB)

# Display the resized image using matplotlib
plt.imshow(img_rgb)
plt.axis('off')  # Hide axes
plt.show()
```

```python
#cropping images
crop = Image2_rgb[75:200, 10:190]
plt.imshow(crop)
```

```python
#resizing images using first way
Image_new = cv.resize(crop, None , fx = 3 , fy = 3)
plt.imshow(Image_new)
```

```python
#Another methid
height = 500
width = 600
dimension = (width, height )
Image_new = cv.resize(crop, dimension )
plt.imshow(Image_new)
```

```python
# new method
desired_width = 500
Aspect_ratio = desired_width / crop.shape[1]
desired_height= int (crop.shape[1] * Aspect_ratio)
dimension = (desired_width , desired_height)
Image_new = cv.resize(crop, dimension , interpolation = cv.INTER_AREA)
plt.imshow(Image_new)
```

```python
 cv.imwrite("/content/lion_processd.jpg", Image_new)
```

```python
#flipping image
Image_new_horizontal = cv.flip(Image_new, 1)
Image_new_vertical = cv.flip(Image_new, 0)
Image_new_both = cv.flip(Image_new, -1)
plt.subplot(131);plt.imshow(Image_new_horizontal);plt.title("Horizontal flip")
plt.subplot(132);plt.imshow(Image_new_vertical);plt.title("vertical flip")
plt.subplot(133);plt.imshow(Image_new_both);plt.title("Both flip")
```
