## Display images

### Subtask:
List the files in the "VQA_RAD Image Folder" and display a few sample images from this folder.


**Reasoning**:
Construct the path to the image folder, list its contents, filter for image files, and display a few sample images as instructed.



```python
import os
from PIL import Image
import matplotlib.pyplot as plt

# Construct the full path to the "VQA_RAD Image Folder"
image_folder_path = os.path.join(path, "VQA_RAD Image Folder")

# List all the files within the image folder
try:
    image_files_in_folder = os.listdir(image_folder_path)
    print(f"Files in '{image_folder_path}':")
    for file in image_files_in_folder:
        print(file)

    # Filter the list to include only image files
    image_files = [f for f in image_files_in_folder if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    # Check if any image files were found
    if len(image_files) > 0:
        print("\nDisplaying some images:")
        # Iterate through a few of them (e.g., the first 3)
        for i, img_file in enumerate(image_files[:3]):
            img_path = os.path.join(image_folder_path, img_file)
            try:
                # Open the image file
                img = Image.open(img_path)
                # Display the image
                plt.imshow(img)
                # Add the image filename as the title
                plt.title(img_file)
                # Turn off the axis
                plt.axis('off')
                # Show the plot
                plt.show()
            except Exception as e:
                # Include error handling
                print(f"Could not display image {img_file}: {e}")
    else:
        print("\nNo image files found in the image folder.")

except FileNotFoundError:
    print(f"Image folder not found at: {image_folder_path}")
except Exception as e:
    print(f"An error occurred while listing files in the image folder: {e}")
```
