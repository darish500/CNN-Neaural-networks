```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing import image
from tensorflow.keras.optimizers import RMSprop, Adam
from tensorflow.keras import layers , models , optimizers
from tensorflow.keras import Model
from tensorflow.keras.applications import resnet50
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Input, Add, Dense, Activation, ZeroPadding2D, BatchNormalization, Flatten, Conv2D, AveragePooling2D, MaxPooling2D, GlobalMaxPooling2D, Dropout
```

```python
import torch
```

```python
train_data_path = "/content/drive/MyDrive/archive (2)/brain_tumor_dataset/brain_tumor_classification/Training"
test_data_path = "/content/drive/MyDrive/archive (2)/brain_tumor_dataset/brain_tumor_classification/Testing"
train_data = ImageDataGenerator(rotation_range=60 ,
                                width_shift_range=0.2,
                                height_shift_range=0.2,
                                fill_mode="nearest",
                                horizontal_flip=True,
                                vertical_flip=True)
test_data= ImageDataGenerator(rotation_range=60 ,
                                width_shift_range=0.2,
                                height_shift_range=0.2,
                                fill_mode="nearest",
                                horizontal_flip=True,
                                vertical_flip=True)
train_data = train_data.flow_from_directory(directory=train_data_path,
                                          target_size=(224,224),
                                          batch_size=50,
                                          class_mode="categorical")
test_data = test_data.flow_from_directory(directory=test_data_path,
                                          target_size=(224,224),
                                          batch_size=50,
                                          class_mode="categorical")
```

```python
# 1. Load the pre-trained ResNet50 model (without the classification head)
from tensorflow.keras.applications.resnet50 import ResNet50
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))

# 2. Freeze the base model layers
for layer in base_model.layers:
    layer.trainable = False

# 3. Create a new model on top
inputs = base_model.input
x = base_model.output
x = layers.GlobalAveragePooling2D()(x)  # Use GlobalAveragePooling2D with ResNet
x = layers.Dense(2048, activation='relu')(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(4, activation='softmax')(x)  # num_classes is the number of your classes
model = models.Model(inputs=inputs, outputs=outputs)
```

```python
optimizer = Adam(learning_rate=0.0001)
model.compile(loss="categorical_crossentropy",optimizer=optimizer,metrics=["accuracy"])
history = model.fit(train_data,epochs=50,
                validation_data=test_data, validation_steps=1500)
```

```python
model.evaluate(test_data)
```
