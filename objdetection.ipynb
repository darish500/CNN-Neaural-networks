```python
# prompt: Check the file and see if it contains image and annotation or label as the case may be

import os

# Define the extracted folder path
extracted_folder_path = '/content/rice_dataset'

# Function to check for images and annotations
def check_dataset_structure(base_path):
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.tiff'] # Add more extensions if needed
    annotation_extensions = ['.xml', '.txt', '.json'] # Add more extensions if needed (e.g., YOLO .txt, Pascal VOC .xml, COCO .json)

    found_images = False
    found_annotations = False

    for root, dirs, files in os.walk(base_path):
        for file in files:
            file_lower = file.lower()
            if any(file_lower.endswith(ext) for ext in image_extensions):
                found_images = True
            if any(file_lower.endswith(ext) for ext in annotation_extensions):
                found_annotations = True

            # If both are found, we can stop searching early
            if found_images and found_annotations:
                return True, True

    return found_images, found_annotations

# Check the extracted dataset folder
images_found, annotations_found = check_dataset_structure(extracted_folder_path)

if images_found:
    print(f"Images found in {extracted_folder_path}")
else:
    print(f"No common image files found in {extracted_folder_path}")

if annotations_found:
    print(f"Annotations (labels) found in {extracted_folder_path}")
else:
    print(f"No common annotation files found in {extracted_folder_path}")

if images_found and annotations_found:
    print("The dataset likely contains both images and annotations for object detection.")
elif images_found and not annotations_found:
    print("Images found, but no common annotation files. This might be a classification dataset or annotations have a different format.")
elif not images_found and annotations_found:
    print("Annotations found, but no common image files. This is unusual for an image dataset.")
else:
    print("Neither common image nor annotation files were found in the extracted directory.")
```
